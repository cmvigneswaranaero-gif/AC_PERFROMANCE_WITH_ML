# ***Function***

```python
# Function to calculate Lift Force
def lift (density, velocity, wing_area, Cl):
    """
    Calculates Lift Force using the formula:
    L = 0.5 * density * velocity^2 * wing_area * Cl
    """
    return 0.5 * density * velocity**2 * wing_area * Cl

"""
If the function call and print statement are on the same line as the return,
which will cause an indentation error
"""
# Call the function
lift_force = lift(1, 200, 25, 1)
print("Lift Force (N):", lift_force)
```

```python
# Function to calculate Reynolds Number
def reynolds_number(rho, V, L, mu):
    return (rho * V * L) / mu

# Function to calculate Dynamic Pressure
def dynamic_pressure(rho, V):
    return 0.5 * rho * V**2


# Call the functions
Re = reynolds_number(1.225,50, 0.5, 1.81e-5)
q = dynamic_pressure(1.225, 50)

# Print results
print("Reynolds Number:", Re)
print("Dynamic Pressure (Pa):", q)
```

# **for loop**

```python
# why - for loop
velocities = [50, 60, 70] # list
q = 2 * velocities
print(q)

# list is multiplied by two times..
# but we need to multiply the individual elements in the list..like array calculation..
# then use for loop

velocities = [50, 60, 70] # list
for V in velocities:
  q = 2 * V
  print(q)
```

```python
velocities = [10, 20, 30]

# 1. Basic for ... in list
for V in velocities:
    print(f"Velocity = {V} m/s")

# 2. Using enumerate() (index + value)
for i, V in enumerate(velocities):
    print(f"Case {i+1}: Velocity = {V} m/s")

# 3. Using range(start, stop, step)
for V in range(10, 40, 10):   # from 10 to 30 with step 10
    print(f"Velocity = {V} m/s")


# thumb rule
# Only index → use range(len(list)) (most common).

# Only value → for value in list:

# Both index + value → enumerate

```

```python

# Case 1: Vary speed, constant density (rho0)

W   = 103047
S   = 47
AR  = 6.5
Cd0 = 0.032
e   = 0.87
TA  = 40298 * 2

k = 1 / (3.14159 * e * AR)

rho = 1.225     # sea level density
V = [30, 40, 50, 60, 70, 80, 90, 100,120, 140, 160, 180, 200, 220, 240, 260, 280, 300]

for v in V:
    Cl = W / (0.5 * rho * v**2 * S)
    Tr = 0.5 * rho * v**2 * S * (Cd0 + k * Cl**2)
    print(f"Speed = {v} m/s --> CL = {Cl:.4f}, Thrust Required = {Tr:.2f} N")
```

***Looping over two lists with zip()***

```python
# Case 2: zip() for element-wise calculation
W   = 103047
S   = 47
AR  = 6.5
Cd0 = 0.032
e   = 0.87
TA  = 40298 * 2

k = 1 / (3.14 * e * AR)
V   = [100, 150, 200, 250, 300, 350]
rho = [1.225, 1.112, 1.007, 0.909, 0.819, 0.736]

for v, r in zip(V, rho):
    Cl = W / (0.5 * r * v**2 * S)
    Tr = 0.5 * r * v**2 * S * (Cd0 + k * Cl**2)
    print(f"Speed = {v} m/s, Density = {r} --> TR = {Tr:.2f} N")

```

# ***Looping Without zip()***

```python
# Case 3: Index-based looping

W   = 103047
S   = 47
AR  = 6.5
Cd0 = 0.032
e   = 0.87
TA  = 40298 * 2

k = 1 / (3.14 * e * AR)
V   = [100, 150, 200, 250, 300, 350]
rho = [1.225, 1.112, 1.007, 0.909, 0.819, 0.736]

for i in range(len(V)):
    v   = V[i]
    r = rho[i]
    Cl = W / (0.5 * r * v**2 * S)
    Tr = 0.5 * r * v**2 * S * (Cd0 + k * Cl**2)
    print(f"Speed = {v} m/s, Density = {r} --> TR = {Tr:.2f} N")

```

# ***NESTED LOOP***

```python
# Case 4: Nested loop for full TR curves

rho = [1.225, 1.112, 1.007, 0.909, 0.819, 0.736]

V = [30, 40, 50, 60, 70, 80, 90, 100,
     120, 140, 160, 180, 200, 220,
     240, 260, 280, 300]

for r in rho:
    for v in V:
        Cl = W / (0.5 * r * v**2 * S)
        Tr = 0.5 * r * v**2 * S * (Cd0 + k * Cl**2)
        print(f"Density = {r} kg/m^3   Speed = {v} m/s --> TR = {Tr:.2f} N")
```

# ***WHILE LOOP***

```python
# Constants
S = 20
CL = 0.5
rho = 1.2
V = 50

# Lift threshold
threshold = 20000

# While loop
while True:
    L = 0.5 * rho * V**2 * S * CL
    print(f"Velocity = {V} m/s --> Lift = {L:.0f} N")

    if L > threshold:
        print("Lift exceeded threshold!")
        break  # exit the loop

    V += 5   # increase velocity each iteration
            # V = V + 5
```

```python
# Constants
rho = 10        # water density, kg/m^3
D = 0.05          # pipe diameter, m
mu = 0.001        # viscosity, Pa.s

# Initial velocity
V = 1

# While loop
while True:
    Re = (rho * V * D) / mu
    print(f"Velocity = {V} m/s --> Reynolds number = {Re:.0f}")

    if Re > 4000:
        print("Flow has become turbulent!")
        break  # stop the loop

    V += 1  # increment velocity
```

```python
# Example: increase velocity until Re > 4000
rho = 10
D = 0.05
mu = 0.001
V = 1

Re = (rho * V * D) / mu
while Re <= 4000:
    print(f"Velocity = {V} m/s --> Re = {Re:.0f}")
    V += 1
    Re = (rho * V * D) / mu
```

```python
# Case 3: All combinations using broadcasting (matrix thrust required table)
# Broadcasting (Vectorized Calculation, Matrix Output)
import numpy as np

W   = 103047
S   = 47
Cd0 = 0.032
AR  = 6.5
e   = 0.87
k   = 1 / (np.pi * e * AR)

rho = np.array([1.225, 1.112, 1.007])   # densities
V   = np.array([60, 80, 100])           # speeds

# Reshape for broadcasting
Cl = W / (0.5 * rho.reshape(3,1) * V**2 * S)
Tr = 0.5 * rho.reshape(3,1) * V**2 * S * (Cd0 + k * Cl**2)

print(Tr)


# NumPy will happily give you a matrix of Lift values for all combinations of ρ and  V, but it won’t automatically label which value belongs to which pair
```

```python
# Case 1: Simple line plot (TR vs V at sea level)
import numpy as np
import matplotlib.pyplot as plt

# Aircraft and aerodynamic parameters
W   = 103047      # Weight (N)
S   = 47          # Wing area (m^2)
AR  = 6.5         # Aspect ratio
Cd0 = 0.032       # Zero-lift drag coefficient
e   = 0.87        # Oswald efficiency factor
k   = 1 / (np.pi * e * AR)

rho = 1.225       # Sea-level density (kg/m^3)

# Speed range
V = np.linspace(30, 300, 50)   # speeds in m/s

# Thrust required calculation
q  = 0.5 * rho * V**2          # dynamic pressure
Cl = W / (q * S)
Tr = q * S * (Cd0 + k * Cl**2)

# Figure 1: Line plot
plt.figure(1)
plt.plot(V, Tr, color='blue', linewidth=2)
plt.xlabel("Speed V (m/s)")
plt.ylabel("Thrust Required TR (N)")
plt.title("Thrust Required vs Speed (Line)")
plt.grid(True)
plt.show()
```

```python
# Case 5: Multiple densities in a loop
import numpy as np
import matplotlib.pyplot as plt

W   = 103047
S   = 47
AR  = 6.5
Cd0 = 0.032
e   = 0.87
k   = 1 / (np.pi * e * AR)

rho = np.array([1.225, 1.112, 1.007, 0.909, 0.819, 0.736])
V   = np.linspace(30, 300, 50)

plt.figure(1)

for r in rho:
    q  = 0.5 * r * V**2
    Cl = W / (q * S)
    Tr = q * S * (Cd0 + k * Cl**2)
    plt.plot(V, Tr, linewidth=2, label=f"rho={r} kg/m³")

plt.xlabel("Speed V (m/s)")
plt.ylabel("Thrust Required TR (N)")
plt.title("Thrust Required vs Speed for different densities")
plt.grid(True)
plt.legend()
plt.show()

```

```python
# Case 6: Nested loop (vary weight and density)
import numpy as np
import matplotlib.pyplot as plt

S   = 47
AR  = 6.5
Cd0 = 0.032
e   = 0.87
k   = 1 / (np.pi * e * AR)

# Two weights (e.g., light and heavy configurations)
Weight   = np.array([80000, 103047])          # N
rho = np.array([1.225, 0.736])          # sea level and 5 km

V = np.linspace(30, 300, 50)

for W in Weight:
    for r in rho:
        q  = 0.5 * r * V**2
        Cl = W / (q * S)
        Tr = q * S * (Cd0 + k * Cl**2)
        plt.plot(V, Tr, label=f"W={W} N, rho={r} kg/m³")

plt.xlabel("Speed V (m/s)")
plt.ylabel("Thrust Required TR (N)")
plt.title("Thrust Required vs Speed for multiple W and rho")
plt.grid(True)
plt.legend()
plt.show()


```

# ***export csv file***

```python
import numpy as np
from openpyxl import Workbook
from openpyxl.utils import get_column_letter
from google.colab import files

# ----- Inputs -----
S_list  = [17, 34, 47, 85, 103]
b_list  = [17.47, 33.0, 54.0, 24.0, 40]
Cd0_list = [0.032, 0.051, 0.10]
e_list   = [0.87, 0.77, 0.93]
Weight = np.arange(50000, 200001, 10000)
rho_vals = [1.225, 1.112, 1.007, 0.819, 0.525, 0.413]
V_vals   = np.linspace(10, 400, 10)
TA0 = 80596.0

# ----- Workbook -----
wb = Workbook()
ws = wb.active
ws.title = "Aero Data"

# Header
ws.append([
    "rho","W","V","S","b","AR","Cd0","e",
    "CL","CD","CL/CD",
    "TR","PR","TA","PA","RC"
])

row = 2  # Excel row index

for S, b in zip(S_list, b_list):
    AR = b*b/S

    for Cd0, e in zip(Cd0_list, e_list):
        k = 1/(np.pi * e * AR)

        for W in Weight:
            for r in rho_vals:
                for V in V_vals:

                    # Compute numeric values (to DISPLAY)
                    q  = 0.5 * r * V**2
                    CL_val = W/(q*S)
                    CD_val = Cd0 + k*CL_val**2
                    TR_val = q*S*CD_val
                    PR_val = TR_val * V
                    TA_val = (r/1.225)*TA0
                    PA_val = TA_val * V
                    RC_val = (PA_val - PR_val)/W
                    CLCD_val = CL_val / CD_val

                    # ----- Write DISPLAY values -----
                    ws.cell(row=row, column=1, value=r)
                    ws.cell(row=row, column=2, value=W)
                    ws.cell(row=row, column=3, value=V)
                    ws.cell(row=row, column=4, value=S)
                    ws.cell(row=row, column=5, value=b)
                    ws.cell(row=row, column=6, value=AR)
                    ws.cell(row=row, column=7, value=Cd0)
                    ws.cell(row=row, column=8, value=e)
                    ws.cell(row=row, column=9, value=CL_val)
                    ws.cell(row=row, column=10, value=CD_val)
                    ws.cell(row=row, column=11, value=CLCD_val)
                    ws.cell(row=row, column=12, value=TR_val)
                    ws.cell(row=row, column=13, value=PR_val)
                    ws.cell(row=row, column=14, value=TA_val)
                    ws.cell(row=row, column=15, value=PA_val)
                    ws.cell(row=row, column=16, value=RC_val)

                    # ----- Attach the formula so Excel shows it in formula bar -----
                    # (Replace A,B,C.. with actual cell coordinates)
                    rho_c = f"A{row}"
                    W_c   = f"B{row}"
                    V_c   = f"C{row}"
                    S_c   = f"D{row}"
                    Cd0_c = f"G{row}"

                    # Store formulas but keep display values
                    ws[f"I{row}"].value  = f"={W_c}/(0.5*{rho_c}*{V_c}*{V_c}*{S_c})"   # CL
                    ws[f"J{row}"].value  = f"={Cd0_c}+({CL_val}*{CL_val})*{k}"          # CD
                    ws[f"K{row}"].value  = f"=I{row}/J{row}"                             # CL/CD
                    ws[f"L{row}"].value  = f"=0.5*{rho_c}*{V_c}*{V_c}*{S_c}*J{row}"       # TR
                    ws[f"M{row}"].value  = f"=L{row}*{V_c}"                               # PR
                    ws[f"N{row}"].value  = f"=({r}/1.225)*{TA0}"                          # TA
                    ws[f"O{row}"].value  = f"=N{row}*{V_c}"                               # PA
                    ws[f"P{row}"].value  = f"=(O{row}-M{row})/{W_c}"                      # RC

                    row += 1

# Save file
filename = "Aero_Results_With_Formulas.xlsx"
wb.save(filename)

# Download in Colab
files.download(filename)
```
