# Data Importing

```
#Importing dataset
aircraft_train=pd.read_csv("https://raw.githubusercontent.com/cmvigneswaranaero-gif/AC_PERFROMANCE_WITH_ML/refs/heads/main/RF_train.csv")
aircraft_test=pd.read_csv("https://raw.githubusercontent.com/cmvigneswaranaero-gif/AC_PERFROMANCE_WITH_ML/refs/heads/main/RF_test.csv")
```

```
features=list(aircraft_train.columns[1:])
X_train=aircraft_train[features]
y_train=aircraft_train['Fatal']
print(features)
```

# Decision Tree

```
###buildng Decision tree on the training data ####
clf = tree.DecisionTreeClassifier(max_depth=6)
clf.fit(X_train,y_train)
```

## Decision Tree Results

```
#####predicting on test data ####
tree_predict_train=clf.predict(aircraft_train[features])
tree_predict=clf.predict(aircraft_test[features])

########Train Data Validation
cm1 = confusion_matrix(aircraft_train[['Fatal']],tree_predict_train)
print("Confusion Matrix for Train data \n", cm1)
total1=sum(sum(cm1))
accuracy_tree=(cm1[0,0]+cm1[1,1])/total1
print("Accuracy of Train data \n", accuracy_tree)

########Test Data Validation
cm1 = confusion_matrix(aircraft_test[['Fatal']],tree_predict)
print("Confusion Matrix for test data \n", cm1)
total1=sum(sum(cm1))
accuracy_tree=(cm1[0,0]+cm1[1,1])/total1
print("Accuracy of test data \n", accuracy_tree)
```

# Random Forest

```
from sklearn.ensemble import RandomForestClassifier
forest=RandomForestClassifier(n_estimators=250, max_features=7, max_depth=10)
# n_estimators - Trees - As high as possible
# max_features - p - low ==>(sqrt(t))
# max_depth - Slightly higher than single tree(50% of single DT)
forest.fit(X_train,y_train)
```

## Random Forest Results

```
#####predicting on test data ####
Forest_predict_train=forest.predict(aircraft_train[features])
Forest_predict=forest.predict(aircraft_test[features])

########Train Data Validation
cm1 = confusion_matrix(aircraft_train[['Fatal']],Forest_predict_train)
print("Confusion Matrix for Train data \n", cm1)
total1=sum(sum(cm1))
accuracy_tree=(cm1[0,0]+cm1[1,1])/total1
print("Accuracy of Train data \n", accuracy_tree)

########Test Data Validation
cm1 = confusion_matrix(aircraft_test[['Fatal']],Forest_predict)
print("Confusion Matrix for test data \n", cm1)
total1=sum(sum(cm1))
accuracy_tree=(cm1[0,0]+cm1[1,1])/total1
print("Accuracy of test data \n", accuracy_tree)
```
