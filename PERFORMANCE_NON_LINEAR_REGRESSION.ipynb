```python
import pandas as pd
import numpy as np
from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import PolynomialFeatures
from sklearn.metrics import r2_score

# ---------------------------------------
# Load your CSV file from GitHub
# ---------------------------------------
url = "https://raw.githubusercontent.com/cmvigneswaranaero-gif/AC_PERFROMANCE_WITH_ML/refs/heads/main/thrust_required_with_formulas.csv"
df = pd.read_csv(url)

# Inputs and Output
X = df[["rho", "V"]]   # features
y = df["TR"]           # target

# =======================================
# 1) LINEAR REGRESSION MODEL
# =======================================
lin_model = LinearRegression()
lin_model.fit(X, y)

# Predictions for linear model
y_linear = lin_model.predict(X)

# R² score for linear model
r2_linear = r2_score(y, y_linear)
print("Linear Regression R² =", r2_linear)

# =======================================
# 2) NON-LINEAR (POLYNOMIAL) REGRESSION
# =======================================
degree = 2        # <-- you can increase to 3 if you want
poly = PolynomialFeatures(degree=degree)
X_poly = poly.fit_transform(X)

poly_model = LinearRegression()
poly_model.fit(X_poly, y)

# Predictions for polynomial model
y_poly = poly_model.predict(X_poly)

# R² score for polynomial regression
r2_poly = r2_score(y, y_poly)
print("Polynomial Regression (deg=", degree, ") R² =", r2_poly)

# =======================================
# Save predictions into dataframe (optional)
# =======================================
df["TR_linear"] = y_linear
df["TR_poly"] = y_poly

print(df[["rho", "V", "TR", "TR_linear", "TR_poly"]].head())
```
