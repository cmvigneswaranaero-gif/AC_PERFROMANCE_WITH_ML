# ***NON LINEAR REGRESSION _SIMPLE***

```python
import pandas as pd
import numpy as np
from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import PolynomialFeatures
from sklearn.metrics import r2_score

# 1. Load your CSV
url = "https://raw.githubusercontent.com/cmvigneswaranaero-gif/AC_PERFROMANCE_WITH_ML/refs/heads/main/Thrust_Required_sample.csv"
df = pd.read_csv(url)


# 2. Extract columns (rho, V, TR)
X = df[['rho', 'V']]
y = df['TR']

# -------------------------------
# LINEAR REGRESSION
# -------------------------------
lin_model = LinearRegression()
lin_model.fit(X, y)

df["TR_linear_pred"] = lin_model.predict(X)     # save predictions
r2_lin = r2_score(y, df["TR_linear_pred"])

print("Linear Regression R² =", r2_lin)

# -------------------------------
# POLYNOMIAL REGRESSION (NON-LINEAR)
# -------------------------------
poly = PolynomialFeatures(degree=2, include_bias=False)
X_poly = poly.fit_transform(X)

poly_model = LinearRegression()
poly_model.fit(X_poly, y)

df["TR_poly_pred"] = poly_model.predict(X_poly)  # save predictions
r2_poly = r2_score(y, df["TR_poly_pred"])

print("Polynomial Regression R² =", r2_poly)

# -------------------------------
# Show first few rows WITH predictions
# -------------------------------
print(df[['rho', 'V', 'TR', 'TR_linear_pred', 'TR_poly_pred']].head())


# ONE SET OF VALUES FOR PREDICTION
# --------------------------------------------------
rho_in = 1.212     # example density
V_in   = 340.0     # example velocity

new_X = np.array([[rho_in, V_in]])
new_X_poly = poly.transform(new_X)

TR_lin  = lin_model.predict(new_X)[0]
TR_poly = poly_model.predict(new_X_poly)[0]


print("Linear Regression TR =", TR_lin)
print("Polynomial Regression TR =", TR_poly)
```

# ***WEIGHT AS INPUT***

```python
import pandas as pd
import numpy as np
from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import PolynomialFeatures
from sklearn.metrics import r2_score

# ---------------------------------------
# 1) Load your CSV file
# ---------------------------------------
url = "https://raw.githubusercontent.com/cmvigneswaranaero-gif/AC_PERFROMANCE_WITH_ML/refs/heads/main/thrust_required_with_formulas.csv"
df = pd.read_csv(url)

# Features: rho, V, W
X = df[["rho", "V", "W"]]

# Targets: TR and RC
y = df[["TR", "RC"]]      # two outputs

# =======================================
# 2) LINEAR REGRESSION (TR and RC together)
# =======================================
lin_model = LinearRegression()
lin_model.fit(X, y)

# Predictions on full data
y_lin = lin_model.predict(X)

# Save predictions into dataframe
df["TR_lin"] = y_lin[:, 0]   # first column = TR
df["RC_lin"] = y_lin[:, 1]   # second column = RC

# R² scores
r2_lin_TR = r2_score(df["TR"], df["TR_lin"])
r2_lin_RC = r2_score(df["RC"], df["RC_lin"])

print("Linear Regression R² (TR) =", r2_lin_TR)
print("Linear Regression R² (RC) =", r2_lin_RC)

# =======================================
# 3) POLYNOMIAL REGRESSION (NON-LINEAR)
# =======================================
degree = 2   # you can change to 3 later if you want
poly = PolynomialFeatures(degree=degree, include_bias=False)

X_poly = poly.fit_transform(X)

poly_model = LinearRegression()
poly_model.fit(X_poly, y)

# Predictions on full data
y_poly = poly_model.predict(X_poly)

df["TR_poly"] = y_poly[:, 0]
df["RC_poly"] = y_poly[:, 1]

# R² scores
r2_poly_TR = r2_score(df["TR"], df["TR_poly"])
r2_poly_RC = r2_score(df["RC"], df["RC_poly"])

print("Polynomial Regression (deg =", degree, ") R² (TR) =", r2_poly_TR)
print("Polynomial Regression (deg =", degree, ") R² (RC) =", r2_poly_RC)

# =======================================
# 4) Show first few rows with predictions
# =======================================
print(df[["rho", "V", "W", "TR", "TR_lin", "TR_poly",
          "RC", "RC_lin", "RC_poly"]].head())

# =======================================
# 5) ONE SET OF VALUES FOR PREDICTION
#    (no user input, just fixed example)
# =======================================
rho_in = 1.112    # example rho
V_in   = 120.0    # example V (m/s)
W_in   = 90000.0  # example W (N)

new_X = np.array([[rho_in, V_in, W_in]])
new_X_poly = poly.transform(new_X)

# Predict TR and RC from both models
TR_RC_lin  = lin_model.predict(new_X)[0]       # [TR, RC]
TR_RC_poly = poly_model.predict(new_X_poly)[0] # [TR, RC]

print("\nExample prediction for rho =", rho_in,
      ", V =", V_in, ", W =", W_in)
print("  Linear   -> TR =", TR_RC_lin[0],  ", RC =", TR_RC_lin[1])
print("  Poly     -> TR =", TR_RC_poly[0], ", RC =", TR_RC_poly[1])
```

# ***DECISION TREE TO IMPROVE R^2***

```python
import pandas as pd
import numpy as np
from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import PolynomialFeatures
from sklearn.metrics import r2_score
from sklearn.tree import DecisionTreeRegressor   # <-- added

# --------------------------------------------------
# 1) Load your CSV (from GitHub)
# --------------------------------------------------
url = "https://raw.githubusercontent.com/cmvigneswaranaero-gif/AC_PERFROMANCE_WITH_ML/refs/heads/main/TR_PR_RC_Output_SAMPLE_2.csv"
df = pd.read_csv(url)
df.columns = df.columns.str.strip()

# --------------------------------------------------
# 2) Inputs (rho, V) and targets (TR, RC)
# --------------------------------------------------
X  = df[['rho', 'V']]
y1 = df['TR']
y2 = df['RC']

# ==================================================
# 3) LINEAR REGRESSION (TR & RC)
# ==================================================
lin_TR = LinearRegression().fit(X, y1)
df['TR_linear_pred'] = lin_TR.predict(X)
print("Linear Regression (TR) R² =", r2_score(y1, df['TR_linear_pred']))

lin_RC = LinearRegression().fit(X, y2)
df['RC_linear_pred'] = lin_RC.predict(X)
print("Linear Regression (RC) R² =", r2_score(y2, df['RC_linear_pred']))

# ==================================================
# 4) POLYNOMIAL REGRESSION (degree = 2)
# ==================================================
poly = PolynomialFeatures(degree=2, include_bias=False)
X_poly = poly.fit_transform(X)

poly_TR = LinearRegression().fit(X_poly, y1)
df['TR_poly_pred'] = poly_TR.predict(X_poly)
print("Polynomial Regression (TR) R² =", r2_score(y1, df['TR_poly_pred']))

poly_RC = LinearRegression().fit(X_poly, y2)
df['RC_poly_pred'] = poly_RC.predict(X_poly)
print("Polynomial Regression (RC) R² =", r2_score(y2, df['RC_poly_pred']))

# ==================================================
# 5) DECISION TREE REGRESSION (simple)
# ==================================================
tree_TR = DecisionTreeRegressor(max_depth=4).fit(X, y1)
df["TR_tree_pred"] = tree_TR.predict(X)
print("Decision Tree (TR) R² =", r2_score(y1, df["TR_tree_pred"]))

tree_RC = DecisionTreeRegressor(max_depth=4).fit(X, y2)
df["RC_tree_pred"] = tree_RC.predict(X)
print("Decision Tree (RC) R² =", r2_score(y2, df["RC_tree_pred"]))

# ==================================================
# Show first few rows
# ==================================================
print(df[['rho','V','TR','TR_linear_pred','TR_poly_pred','TR_tree_pred',
          'RC','RC_linear_pred','RC_poly_pred','RC_tree_pred']].head())

# ==================================================
# 6) ONE SET OF GIVEN VALUES TO PREDICT TR & RC
# ==================================================
rho_in = 1.112
V_in   = 120.0

new_X = np.array([[rho_in, V_in]])
new_X_poly = poly.transform(new_X)

TR_lin  = lin_TR.predict(new_X)[0]
TR_poly = poly_TR.predict(new_X_poly)[0]
TR_tree = tree_TR.predict(new_X)[0]

RC_lin  = lin_RC.predict(new_X)[0]
RC_poly = poly_RC.predict(new_X_poly)[0]
RC_tree = tree_RC.predict(new_X)[0]

print("\nFor rho =", rho_in, "and V =", V_in)
print("  Linear     TR =", TR_lin)
print("  Polynomial TR =", TR_poly)
print("  Tree       TR =", TR_tree)

print("-------------------------------------")

print("  Linear     RC =", RC_lin)
print("  Polynomial RC =", RC_poly)
print("  Tree       RC =", RC_tree)
```

# ***VERY COMPLICATE DATA SET***

```python
import pandas as pd
import numpy as np
from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import PolynomialFeatures
from sklearn.metrics import r2_score

# ---------------------------------------
# Load your CSV file from GitHub
# ---------------------------------------
url = "https://raw.githubusercontent.com/cmvigneswaranaero-gif/AC_PERFROMANCE_WITH_ML/refs/heads/main/thrust_required_with_formulas.csv"
df = pd.read_csv(url)

# Inputs and Output
X = df[["rho", "V"]]   # features
y = df["TR"]           # target

# =======================================
# 1) LINEAR REGRESSION MODEL
# =======================================
lin_model = LinearRegression()
lin_model.fit(X, y)

# Predictions for linear model
y_linear = lin_model.predict(X)

# R² score for linear model
r2_linear = r2_score(y, y_linear)
print("Linear Regression R² =", r2_linear)

# =======================================
# 2) NON-LINEAR (POLYNOMIAL) REGRESSION
# =======================================
degree = 2        # <-- you can increase to 3 if you want
poly = PolynomialFeatures(degree=degree)
X_poly = poly.fit_transform(X)

poly_model = LinearRegression()
poly_model.fit(X_poly, y)

# Predictions for polynomial model
y_poly = poly_model.predict(X_poly)

# R² score for polynomial regression
r2_poly = r2_score(y, y_poly)
print("Polynomial Regression (deg=", degree, ") R² =", r2_poly)

# =======================================
# Save predictions into dataframe (optional)
# =======================================
df["TR_linear"] = y_linear
df["TR_poly"] = y_poly

print(df[["rho", "V", "TR", "TR_linear", "TR_poly"]].head())
```
